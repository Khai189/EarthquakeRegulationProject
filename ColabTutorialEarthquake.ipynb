## Introduction

Before 2019, Oklahoma suffered over a [hundred 3.0+ magnitude earthquakes a year](https://www.krwg.org/krwg-news/2023-07-13/oklahomas-shaky-ground-tamed-by-industry-regulatory-cooperation). In 2015, the state saw an upwards of 900 3.0+ magnitude earthquakes. After 2019, this number significantly dropped to below one hundred and stayed that way for the years following. You are probably wondering: what caused this massive drop in the number of earthquakes?

One potential answer is the increased regulation on wastewater injection policies established in 2019. Wastewater injection a process where manufacturing companies inject liquid waste generated from their production deep into the ground. Now the question is: is there any strong evidence to show that waste water injections caused the sharp decrease in earthquake frequency?

 Experts in geology and environmental science could possibly establish a link, but not everyone is an expert in these fields. Instead, we can rely on statistics, a discipline that allows us to take trends we see in data and thoroughly argue if the observations have underlying patterns based on mathematical frameworks.
